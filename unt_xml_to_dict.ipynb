```python
import xmltodict
 
#open the file
file = open("untl_sample.xml","r")

xml_content= file.read()
#print("XML content is:")
#print(xml_content)

unt_xml = xmltodict.parse(xml_content)
#print(unt_xml['ns0:record'])
# All item metadata
#print(unt_xml['ns0:record']['ns0:metadata']['ns1:metadata'])

item_meta = unt_xml['ns0:record']['ns0:metadata']['ns1:metadata']

print('TITLE')
title = item_meta['ns1:title']
print(title['#text'])
print('DATE')
date = item_meta['ns1:date']
print(date)

print('DESCRIPTION')
description = item_meta['ns1:description']
print(description)

print('SUBJECTS')
subjects = item_meta['ns1:subject']
subject_list = []
for subject in subjects:
    #print(subject)
    subject_list.append(subject['#text'])
print(subject_list)


print('COVERAGES')
coverages = item_meta['ns1:coverage']
for coverage in coverages:
    #print(coverage)
    if coverage['@qualifier'] == 'placeName':
        placeName = coverage['#text']
        print('placeName:', placeName)

print('CITATION')
citation = item_meta['ns1:citation']
print(citation)

print('IDENTIFIERS')
identifiers = item_meta['ns1:identifier']
for identifier in identifiers:
    print(identifier)

print('UNT META')
unt_metas = item_meta['ns1:meta']
for unt_meta in unt_metas:
    print(unt_meta)

# NOTE - not parsed:
# <ns1:primarySource>
# <ns1:creator qualifier="pht">
# <ns1:language>
# <ns1:collection>
# <ns1:institution>
# <ns1:rights qualifier="license">
# <ns1:resourceType>
# <ns1:format>
```
